## Q1. Arithmetic and Assignment Operators

```python
# defining two numeric variables
a = 15
b = 4

# performing all arithmetic operations
print("a =", a, "and b =", b)
print()
print("Addition:", a + b)
print("Subtraction:", a - b)
print("Multiplication:", a * b)
print("Division:", a / b)
print("Floor Division:", a // b)
print("Modulus:", a % b)
print("Exponentiation:", a ** b)
```

```python
# showing assignment operators
a = 15
print("Initial value of a:", a)

a += 5
print("After a += 5 :", a)

a -= 3
print("After a -= 3 :", a)

a *= 2
print("After a *= 2 :", a)

a /= 4
print("After a /= 4 :", a)
```

## Q2. Comparison and Logical Operators

```python
# taking two integer inputs from the user
x = int(input("Enter first number: "))
y = int(input("Enter second number: "))

# comparison operators
print("\n--- Comparison Results ---")
print(f"{x} == {y} :", x == y)
print(f"{x} != {y} :", x != y)
print(f"{x} > {y}  :", x > y)
print(f"{x} < {y}  :", x < y)
print(f"{x} >= {y} :", x >= y)
print(f"{x} <= {y} :", x <= y)

# logical operators
print("\n--- Logical Operators ---")
print("x > 0 and y > 0 :", x > 0 and y > 0)
print("x > 0 or y < 0  :", x > 0 or y < 0)
print("not(x == y)     :", not(x == y))
```

## Q3. Bitwise Operators

```python
# two integers for bitwise demonstration
p = 12   # binary: 1100
q = 10   # binary: 1010

print("p =", p, "(binary:", bin(p), ")")
print("q =", q, "(binary:", bin(q), ")")
print()

print("Bitwise AND  (p & q)  :", p & q,  "=>", bin(p & q))
print("Bitwise OR   (p | q)  :", p | q,  "=>", bin(p | q))
print("Bitwise XOR  (p ^ q)  :", p ^ q,  "=>", bin(p ^ q))
print("Bitwise NOT  (~p)     :", ~p)
print("Left Shift   (p << 1) :", p << 1, "=>", bin(p << 1))
print("Right Shift  (p >> 1) :", p >> 1, "=>", bin(p >> 1))
```

## Q4. Identity and Membership Operators

```python
# creating a list of numbers
numbers = [2, 5, 8, 11, 14, 17, 20]

# identity operators
var1 = numbers
var2 = [2, 5, 8, 11, 14, 17, 20]  # same values but different object

print("--- Identity Operators ---")
print("var1 is numbers     :", var1 is numbers)      # True - same object
print("var2 is numbers     :", var2 is numbers)      # False - different object
print("var2 is not numbers :", var2 is not numbers)  # True

# membership operators
print("\n--- Membership Operators ---")
print("8 in numbers         :", 8 in numbers)
print("99 in numbers        :", 99 in numbers)
print("5 not in numbers     :", 5 not in numbers)
print("100 not in numbers   :", 100 not in numbers)
```

## Q5. String Operations

```python
# creating a string variable
sentence = "data science is the future of technology"

# length of the string
print("Original string :", sentence)
print("Length          :", len(sentence))

# uppercase and lowercase
print("Uppercase       :", sentence.upper())
print("Lowercase       :", sentence.lower())

# find a word using 'in'
print("\n'future' in sentence :", 'future' in sentence)
print("'coding' in sentence :", 'coding' in sentence)

# replace a word
new_sentence = sentence.replace("future", "backbone")
print("\nAfter replace   :", new_sentence)

# slicing
print("\nSlice [0:12]    :", sentence[0:12])

# reverse using slicing
print("Reversed string :", sentence[::-1])
```
